```python
import importlib
import pandas as pd

import pvpcalc.pipeline as pipeline

pipeline = importlib.reload(
    pipeline
)


audit = await pipeline.audit_spec(
    "Priest",
    "Discipline",
)


print("=" * 90)
print("PRIEST / DISCIPLINE PvP AUDIT")
print("=" * 90)

print(
    "Tree build:          ",
    audit.tree_build,
)

print(
    "Drustvar build:      ",
    ", ".join(
        audit.drustvar_builds
    ),
)

print(
    "Talent entries:      ",
    audit.total_entries,
)

print(
    "Unique spells:       ",
    audit.unique_spells,
)

print(
    "Wowhead candidates:  ",
    len(
        audit.wowhead_candidate_ids
    ),
)

print(
    "Drustvar candidates: ",
    len(
        audit.drustvar_candidate_ids
    ),
)

print(
    "Union candidates:    ",
    len(
        audit.candidate_ids
    ),
)

print(
    "Fetch errors:        ",
    len(
        audit.fetch_errors
    ),
)

print(
    "Unresolved:          ",
    len(
        audit.unresolved_rows
    ),
)


modified_df = pd.DataFrame(
    audit.modified_effect_rows
)

display(
    modified_df[
        [
            "tree_type",
            "hero_tree",
            "talent_name",
            "spell_id",
            "effect_index",
            "effect_text",
            "base_value",
            "pvp_multiplier",
            "pvp_value",
            "drustvar_matched",
            "match_reason",
            "wowhead_multiplier",
            "drustvar_multiplier",
            "confidence",
        ]
    ]
    .sort_values(
        [
            "tree_type",
            "hero_tree",
            "talent_name",
            "effect_index",
        ],
        na_position="first",
    )
    .reset_index(drop=True)
)


if audit.unresolved_rows:

    print("\nUNRESOLVED")

    display(
        pd.DataFrame(
            audit.unresolved_rows
        )
    )
```
